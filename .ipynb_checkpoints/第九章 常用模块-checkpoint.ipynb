## 9.1 sys模块

sys模块代表了Python系统主要是解释器的相关信息设置。  .
sys模块的参考页面在：https://docs.python.org/3/library/sys.html

下面的命令给出了sys模块中的全部共有变量（包含变量，函数）。

```python
import sys
[e for e in dir(sys) if not e.startswith('_')]
```

上述变量不必全部记忆，通常我们只需要了解常用的功能即可。  

+ sys.argv: 获取运行Python程序的命令行参数
+ sys.byteorder: 显示本地字节符的指示符。如果本地字节符是大端模式，返回big,否则返回little.字节序在网络传输和数据存储时非常重要。
+ sys.copywrite:该属性返回与解释器相关的版权信息。
+ sys.executeable:该属性返回解释器在磁盘的存放路径。
+ sys.exit():通过引发SystemExit异常来退出程序。
+ sys.getrefcount(obj):返回指定对象的引用计数。当obj对象的引用计数为0时，系统回收该对象。
+ sys.modules:返回模块名和载入模块对应关系的字典。
+ sys.path:该属性指定Python解释器查找模块的路径列表。通常修改该属性以动态增加模块的路径。
+ sys.version:返回当前Python解释器的版本信息。
+ sys.winver:返回当前Python解释器的主版本号。

```python
#sys模块使用示例
import sys
print(sys.byteorder)
print(sys.copyright)
print(sys.executable)
print(sys.version)
print(sys.winver)
print(sys.path)
```

查看os模块可以参考：https://docs.python.org/3/library/os.html

使用下面的命令可以查看模块内全部属性和函数。

了解常用属性：

+ os.name:filter回导入依赖模块的操作系统名称。
+ os.environ:filter回当前系统上所有环境变量组成的字典。

## 9.3 random模块

random模块主要包含生成伪随机数的各种功能变量和函数。

下列命令可以展示该模块全部属性和函数：

```python
import random
print(random.__all__)
```

random模块的参考页面在：https://docs.python.org/3/library/random.html

random模块提供了如下常用函数:
+ random.seed(a=None,version=2):指定种子来初始化伪随机数生成器。
+ random.randrange(start,stop,[step]):返回从start开始到stop结束，步长为step的随机数。
+ random.randint(a,b):生成一个范围在[a,b]之间的随机数。相当于randrange(a,b+1)
+ random.choice(seq):从seq中随机抽取一个元素。如果seq为空，则引发IndexError异常。
+ random.choices(seq,weights=None,*,cum_weights=None,k=1):从seq中抽取k个元素，还可以指定其被抽中的可能性，用weights表示。
+ random.shuffle(x[,random]):对x序列执行洗牌操作。
+ random.sample(population,k):从population序列中随机抽取k个独立的元素。
+ random.random():生成一个[0.0,1.0)之间的伪随机数

```python
#上述部分命令示例：
import random
print(random.random())
print(random.uniform(2.5,10.0))  
print(random.randrange(10))
print(random.randrange(0,21,2))   #随机偶数
```

```python
#生成由字母，数字组成的6位验证码
import random
if __name__=='__main__':
    verificationcode=''
    for num in range(6):
        index=random.randrange(0,6)
        print(num,index,'\n')
        if index!=num and index+1!=num:
            verificationcode += chr(random.randint(97,122))  #小写字母
        elif index+1 ==num:
            verificationcode+=chr(random.randint(65,90))     #大写字母
        else:
            verificationcode += chr(random.randint(48,57))     #数字
    print("6位验证码为：", verificationcode)
```

## 9.4 time模块

Python有很多处理日期和时间的方法，其中转换日期格式是最为常见的。Python提供了time和calendar模块用于格式化日期和时间。本节将针对这两个模块的函数进行详细地介绍。

time表现方式有三种：

1．时间戳（timestamp）的方式
时间戳表示的是从1970年1月1日00:00:00开始按秒计算的偏移量。返回的是float类型，返回时间戳的函数有time()、clock()。

2．元组（struct_time）方式
struct_time元组共有9个元素，返回struct_time的函数主要有gmtime()，localtime()，strptime()。

![](.\图片\time_struct.png)

3．格式化字符串（format time）方式
格式化时间，已格式化的结构使时间更具可读性。包括自定义格式和固定格式，比如“2018-5-11”。


```python
import time
t=time.localtime()

print(t)
print(t.tm_year) 
print(type(t.tm_sec))
print("*"*70)
print("完整星期名称：",time.strftime("%A ",t))
print("完整月份名称：",time.strftime("%B ",t))
print("一天中的第几个小时：",time.strftime("%H ",t))
print("分钟数：",time.strftime("%M ",t))
print("秒数：",time.strftime("%S ",t))
print("本地相应时间：",time.strftime("%X ",t))
print("完整年份",time.strftime("%Y ",t))
print("时区名称",time.strftime("%Z ",t))
```

## 9.5 字符串

对于在编程过程中对字符串的处理操作较多，本节主要针对字符串的常见操作比如连接、截取、检索等操作，然后借助正则表达式实现字符串的复杂处理。


### 9.5.1 字符串的编码转换模块的定义与使用

字符串是由一系列字符组成的不可变序列容器，在计算机中存储的是字符编码值。通常字节（Byte）是计算机存储的最小单位，等于8位bit。而字符是指单个的数字，文字，符号。


在Python中，有两种常用的字符串类型分别为str和bytes。其中，str表示Unicode字符（ASCII或者其他）；bytes表示二进制数据（包括编码的文本）。这两种类型的字符串不能拼接在一起使用。


1  字符串的编码

用于将字符串转换为二进制数据（即bytes），也称为“编码”。encode()方法为str对象的方法，其语法格式如下：  

str.encode([encoding="utf-8"][,errors="strict"])

参数说明如下：
- str：表示要进行转换的字符串。
- encoding="utf-8":可选参数，用于指定进行转码时采用的字符编码，默认为UTF-8,如果想使用简体中文，也可以设置为gb2312。

```python
str='不忘初心'
byte=str.encode('GBK')
print("原字符串：",str)
print("转换后：",byte)
```

2 字符串的解码

用于将二进制数据转换为字符串，称为“解码”。decode()方法为bytes对象的方法，即将使用encode()方法转换的结果再转换为字符串。


bytes.decode([encoding="utf-8"][,errors="strict"])

说明：   
bytes:表示要进行转换的二进制数据，通常是encode()方法转换的结果。   
encoding="utf-8"：可选参数，用于指定进行解码时采用的字符编码，默认为UTF-8,如果想使用简体中文，也可以设置为gb2312。当只有这一个参数时，也可以省略前面的“encoding=”，直接写编码。


```python
bytes=b'\xb2\xbb\xcd\xfc\xb3\xf5\xd0\xc4'
str=bytes.decode('GBK')
print("解码前：",bytes)
print("解码后：",str)
```

### 9.5.2 字符串的常见操作

在Python开发过程中，为了实现某项功能，经常需要对某些字符串进行特殊处理，如连接字符串、截取字符串、格式化字符串等。字符串的常用操作方法有查找、修改和判断三大类。


#### 1.字符串查找

字符串查找常用的方法：
- count():   count()方法用于检索指定字符串在另一个字符串中出现的次数。如果检索的字符串不存在，则返回0，否则返回出现的次数。
- find():    检测某个子串是否包含在这个字符串中，如果在返回这个子串开始的位置下标，否则则返回-1。
- index():   index()方法与find()方法类似，也是用于检索是否包含指定的子字符串，返回首次出现子字符串的位置索引。如果没有找到，则抛出异常。
- startwith():该方法用于检索字符串是否以指定子字符串开头。如果是，则返回True,否则返回False。
- endwith():  该方法用于检索字符串是否以指定子字符串结尾。如果是，则返回True,否则返回False。

```python
#count()和Index()方法示例
str="this is string example....wow!!!"
sub="i"
print("str.count(sub,0,40)：",str.count(sub,0,40))
print("str.count(sub,1,40)：",str.count(sub,1,40))
print("str.count(sub,2,40)：",str.count(sub,2,40))
print("str.count(sub,3,40)：",str.count(sub,3,40))
print("str.count(sub,4,40)：",str.count(sub,4,40))
sub="wow"
print("str.count(sub)：",str.count(sub))

info='abca'
#从下标0开始，查找在字符串中第一个出现的子串
print(info.find('a'))
#从下标1开始，查找在字符串中第一个出现的子串
print(info.find('a',1)) 
#查找不到，返回-1
print(info.find('3'))   
```

```python
#index()，startswith()，endswith()方法示例

url="http://www.baidu.com"

if url.startswith("http://"):
    print("使用的是http协议")

fllename="flower.jpg"
if fllename.endswith(".jpg"):
    print("这是一张图片")

print(url.index('id'))    
print(url.index("sa")) 
```

此外，下列方法用于判断字符串构成：
- isdigit() 方法用于检测字符串是否只由数字组成。
- isalpha() 方法检测字符串是否只由字母或文字组成。
- isalnum() 方法检测字符串是否由字母和数字组成。
- isspace() 方法用于检测字符串是否只由空白字符组成。

#### 2.字符串修改

常见字符串修改操作：    
a. 字符串替换  
b. 字符串分隔  
c. 字符串合并  
d. 去除空格和特殊字符

a.字符串替换：  

replace()方法把字符串中的old（旧字符串）替换成 new(新字符串)，如果指定第三个参数max，则替换不超过 max 次。  
该方法生成新的字符串，原字符串不受影响

```python
str.replace(old, new[, max])
```

```python
str="this is string example....wow!!!this is really string"
print(str.replace("is","was"))
print(str.replace("is","was",3))
print(str)   #原字符串没有改变
```

b.字符串分隔

字符串对象的split()方法可以实现字符串分割。即把一个字符串按照指定的分隔符切分为字符串列表。该列表的元素中，不包括分隔符。  
split()方法的语法格式：str.split(sep,maxsplit)

c.字符串合并

合并字符串将多个字符串采用固定的分隔符连接在一起。字典只对键进行连接。  
os.path.join()是将多个路径组合后返回。  
其语法格式：strnew = string.join(iterable)

```python
url="www.ncepu.educ.cn"
print("使用默认分隔符： ",url.split())
print("以.为分隔符： ",url.split('.'))
print("以.为分隔符，分割0次： ",url.split('.',0))
print("以.为分隔符，分割1次： ",url.split('.',1))
print("以.为分隔符，分割2次： ",url.split('.',2))
print("以.为分隔符，分割2次，并取序号为1的项： ",url.split('.',2)[1])
#（等同于不加次数参数）
print("以.为分隔符，分割最多次 ",url.split('.',-1))
```

```python
#对序列进行操作
a=['1aa','2bb','3cc','4dd','5ee']
print(''.join(a))
print(' '.join(a))
print('-'.join(a))
#对字符串进行操作
b='hello world'
print(' '.join(b))
print('-'.join(b))
#对元组进行操作
c=('aa','bb','cc','dd','ee')
print(' '.join(c))
print('-'.join(c))
#对字典进行操作
d={'name1':'zhangsan','name2':'lisi','name3':'wangwu'}
print(' '.join(d))
print('-'.join(d))
```

使用“+”运算符可完成对多个字符串的连接，“+”运算符可以连接多个字符串并产生一个字符串对象。


```python
cn = "生活就像一盒巧克力，没人知道你会得到什么。"
en = " Life is like a box of chocolates. No one knows what you're going to get."
c_e=cn+en
print(c_e)
```

d.去除空格和特殊字符

用户在输入数据时，可能会无意中输入多余的空格，或在一些情况下，字符串前后不允许出现空格和特殊字符，此时就需要去除字符串中的空格和特殊字符（制表符“\t”、回车符“\r”、换行符“\n”等）。

- strip()方法用于去掉字符串左、右两侧的空格和特殊字符

- rtrip()方法用于去掉字符串左、右两侧的空格和特殊字符

e.字母大小写转换

在Python中，字符串对象提供了lower()方法和upper()方法进行字母的大小写转换。比如应用于验证码的输入，大小写都可以验证。

- lower()方法用于将字符串中的全部大写字母转换为小写字母。如果字符串中没有应该被转换的字符，则将原字符串返回；否则将返回一个新的字符串，将原字符串中每个该进行小写转换的字符都转换成等价的小写字符

f.截取字符串

截取字符串，可以采用切片方法实现。通过切片方法截取字符串的语法格式如下：

#### 3.字符串格式化

格式化字符串是指先指定一个模板，即在模板中预留几个占位符，然后再根据需要填上相应的内容。Python推荐使用format()格式化方法。

<模板字符串>.format(<逗号分隔的参数>)

其中，模版字符串是一个由字符串和占位符组成的字符串。占位符用{}表示。

```python
"{}曰：学而时习之，不亦说乎！".format('孔子')
```

模版字符串由多个占位符，其占位符内没有指定序号，则按照其出现的顺序显示。

```python
"{}曰：学而时习之，不亦{}！".format('孔子','说乎')
```

占位符可以使用从0开始的序号，以指定其出现的位置。

```python
"{1}曰：学而时习之，不亦{0}！".format('说乎','孔子')
```

如果占位符数量予与参数数量不一致，则必须使用序号，否则会报错。

```python
"{}曰：学而时习之，不亦{}！".format('孔子')
```

```python
"论语是{0}弟子所做。{0}曰：学而时习之，不亦说乎！".format('孔子')
```

```python
#位置参数
stu=['maomao',18]
print('the boy is {},age :{}'.format('maomao',18))
print('the boy is {1},age :{0}'.format(18,'maomao'))
print('the boy is {},age :{}'.format(*stu))
```

```python
#使用索引
stu=['maomao',18]
str0='the boy is {0[0]},age :{0[1]}'.format(stu)
print(str0)
```

```python
#关键字参数
tu={'name':'maomao','age':18}
print('the boy is {name},age :{age}'.format(name='maomao',age=18))
```

模版字符串中可以指定格式：

{<序号>：<填充><对齐><宽度><千分位分隔符><精度><类型>}

上面6个控制标记可分为2组。

第一组是<填充><对齐><宽度>，主要用于对显示格式的规范。

- 填充：指定空白处填充的字符。     
- 对齐：指定数据的对齐方式，具体的对齐方式如下： 
    - '>':右对齐  
    - '<':左对齐 ,默认值
    - '=':右对齐，同时将符号放置在填充内容最右侧，该选项只对数字有效
    - '^':数据居中，此选项与width参数共同使用
- 宽度：指定输出数据时所占的宽度。

注意，
- 当填充字符为逗号(,)且作用于整数或浮点数时，该整数（或浮点数）会以逗号分隔的形式输出，例如（1000000会输出 1,000,000）。
- 符号：指定有无符号数，在宽度前使用，此参数的值以及对应的含义如下
    - '+' :正数前加正号，负数前加负号。
    - '-' :正数前不加正号，负数前加负号。
    - ' ' :正数前加空格，负数前加负号。
    - '#' :对于二进制数、八进制数和十六进制数，使用此参数，各进制数前会分别显示 0b、0o、0x前缀；反之则不显示前缀。

```python
s='华北电力大学'
```

数字表示宽度，实际宽度大于设定宽度，则使用实际值。

```python
'{:<30}'.format(s)
```

```python
'{:>30}'.format(s)
```

```python
'{:^30}'.format(s)
```

```python
'{:#^30}'.format(s)
```

格式控制也可以用变量表示

```python
x='*'
```

```python
#制定字符填充
'{0:{1}^20}'.format(s,x)
```

```python
'{0:{1}^{2}}'.format(s,x,22)
```

```python
y='>'
'{0:{1}{3}{2}}'.format(s,x,22,y)
```

```python
第二组<千分位分隔符><精度><类型>主要用于对数值本身的规范。
```

- 类型:指定输出数据的具体类型。
    - s:字符串
    - d:十进制整数
    - c:将十进制整数转换为对于Unicode字符
    - e/E:转换为科学计数法后输出
    - g/G:自动在e和f切换
    - b:将十进制转为为2进制，再输出
    - 0:将十进制转为为8进制，再输出
    - x/X:将十进制转为为16进制，再输出
    - f/F:转换位浮点数，再输出
    - %:显示百分比


```python
'{:-^20}'.format(123456789)
```

```python
'{:-^20,}'.format(123456789)
```

精度有由小数点（.）开头。对于浮点数，精度小数输出的有效位数。对于字符串精度表示输出的最大长度。

```python
'{:.2f}'.format(12345.6789)
```

```python
'{:20.2f}'.format(12345.6789)
```

```python
'{:.5}'.format(s)
```

```python
'{:>.15}'.format(s)
```

建议掌握以下格式

```python
#输出小数点后两位
'{0:.2f}'.format(3.14159)
```

```python
#输出字符串的前4位
'{:.4}'.format(s)
```

```python
#居中并填充
'{:-^10}'.format(s)
```

## 9.6 正则表达式

  正则表达（Regular Expression）就是通过一个文本模式来匹配一组符合条件的字符串，即对目标字符串进行过滤操作。这个文本模式是由一些字符和特殊符合组成的字符串，它们描述了模式的重复或表述多个字符，所以正则表达式能按着某种模式匹配一系列有相似特征的字符串。

Python提供了re模块，用于实现正则表达式的操作。在实现时，可以使用re模块提供的方法（如search()、match()、 findall()等）进行字符串处理，也可以先使用re模块的compile()方法将模式字符串转换为正则表达式对象，然后再使用该正则表达式对象的相关方法来操作字符串。


re模块在使用时，需要先应用import语句引入，具体代码：

import re

### 9.6.1 匹配字符串match

match()方法用于指定文本模式和待匹配的字符串。从字符串的开始处进行匹配，如果在起始位置匹配成功，则返回match对象，否则返回None。


re.match(pattern, string, [flags])


pattern：表示模式字符串，由要匹配的正则表达式转换而来。  
string：表示待匹配的字符串。  
flags：可选参数，表示标志位，用于控制匹配方式，如是否区分字母大小写。常用的标志如表所示。


在使用match()方法时，如果匹配成功，则match方法返回Match对象，然后可以调用对象中的group方法获取匹配成功的字符串，如果文本模式就是一个普通的字符串，那么group方法返回的就是文本模式本身。


通常情况下，match()方法返回的对象可以用group()或groups()方法匹配对象函数来获取匹配表达式。
group(num=0):匹配整个表达式的字符串，group()方法可以一次输入多个组号，在这种情况下它将返回一个包含这些组所对应的元组。
groups():返回一个包含所有小组字符串的的元组。


```python
import re
strurl="www.ncepu.edu.cn"
m=re.match("www.",strurl)
if m is not None:
    print(m.group())
print(m)

print('-'*20) #--------------
m=re.match("ncepu",strurl)
if m is not None:
    print(m.group())
print(m)
```

### 9.6.2 正则表达式

正则表达（Regular Expression）就是通过一个文本模式来匹配一组符合条件的字符串，即对目标字符串进行过滤操作。这个文本模式是由一些字符和特殊符合组成的字符串，它们描述了模式的重复或表述多个字符，所以正则表达式能按着某种模式匹配一系列有相似特征的字符串。

正则表达式常见符号包括：

1.字符组：


[字符组]要考虑的是在同一个位置上可以出现的字符的范围。在同一个位置可能出现的各种字符组成了一个字符组，在正则表达式中用[]表示。字符分为很多类，比如数字、字母、标点等等。应用如表所示。

![](.\正则字符组.png)

2.元字符

正则表达式语言由两种基本字符类型组成：原义（正常）文本字符和元字符。元字符使正则表达式具有处理能力。所谓元字符就是指那些在正则表达式中具有特殊意义的专用字符，可以用来规定其前导字符（即位于元字符前面的字符）在目标对象中的出现模式。


![](.\正则元字符.png)

3.量词


多次显示需要用到量词

![](.\正则量词.png)

量词还可以细分为：贪婪，惰性，支配性。

- 贪婪量词：先看整个字符串是不是一个匹配。如果没有发现匹配，它去掉最后字符串中的最后一个字符，并再次尝试。
- 惰性量词: 先看字符串中的第一个字母是不是一个匹配。如果单独这一个字符还不够，就读入下一个字符，组成两个字符的字符串。如果还是没有发现匹配，惰性量词继续从字符串添加字符直到发现一个匹配或者整个字符串都检查过也没有匹配。 
- 支配量词：只尝试匹配整个字符串。如果整个字符串不能产生匹配，不做进一步尝试，支配量词其实简单的说，就是一刀切。


```python
import re

# 使用贪婪量词
m = re.search(r'\d{5,8}', '87654321')  # 出现数字8次 
print(m)  # 匹配字符'87654321'

# 使用惰性量词
m = re.search(r'\d{5,8}?', '87654321')  # 出现数字5次 
print(m)  # 匹配字符'87654'
```

### 9.6.3 搜索字符串search

re.search(pattern, string, flags=0)


pattern：匹配的正则表达式  
string ：要匹配的字符串。  
Flags：标志位，用于控制正则表达式的匹配方式，如：是否区分大小写，多行匹配等等。  
匹配成功则返回一个匹配的对象，否则返回None。  


```python
import re
line="Cats are smarter than dogs"
searchObj=re.search('(.*) are (.*?) .*',line,re.M|re.I)
if searchObj:
    print("searchObj.group():",searchObj.group())
    print("searchObj.group(1):",searchObj.group(1))
    print("searchObj.group(2):",searchObj.group(2))
else:
    print("Nothing found!!")
```

```python
#match()与search()方法的区别
import re
line="Cats are smarter than dogs"
matchObj=re.match(r'dogs',line,re.M|re.I)
if matchObj:
    print("match-->matchObj.group():",matchObj.group())
else:
    print("No match")

matchObj=re.search(r'dogs',line,re.M|re.I)
if matchObj:
    print("match-->search.group():",matchObj.group())
else:
    print("No match")
```
